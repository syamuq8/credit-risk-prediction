# Credit Risk Prediction System - Exploratory Data Analysis (EDA)

This notebook performs exploratory data analysis on the credit risk dataset to understand distributions, relationships, and statistical traits of the applicant data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

## 1. Load the Credit Risk Dataset

```python
df = pd.read_csv('../dataset/credit_risk_dataset.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 2. General Dataset Statistics & Quality

```python
df.info()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("\nMissing values:")
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df.describe()
```

## 3. Target Variable Analysis (`Risk Level`)

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='Risk Level', data=df, palette=['#10b981', '#f59e0b', '#f43f5e'])
plt.title('Distribution of Risk Levels')
plt.xlabel('Risk Level')
plt.ylabel('Count')
plt.show()
```

## 4. Distibutions of Key Numerical Features

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(df['Credit Score'].dropna(), kde=True, ax=axes[0,0], color='blue')
axes[0,0].set_title('Credit Score Distribution')

sns.histplot(df['Annual Income'], kde=True, ax=axes[0,1], color='green')
axes[0,1].set_title('Annual Income Distribution')

sns.histplot(df['Loan Amount'], kde=True, ax=axes[1,0], color='orange')
axes[1,0].set_title('Loan Amount Distribution')

sns.histplot(df['Debt-to-Income Ratio'].dropna(), kde=True, ax=axes[1,1], color='red')
axes[1,1].set_title('Debt-to-Income Ratio Distribution')

plt.tight_layout()
plt.show()
```

## 5. Bivariate & Multivariate Analysis

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Risk Level', y='Credit Score', data=df, palette=['#10b981', '#f59e0b', '#f43f5e'])
plt.title('Credit Score vs. Risk Level')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Risk Level', y='Debt-to-Income Ratio', data=df, palette=['#10b981', '#f59e0b', '#f43f5e'])
plt.title('DTI Ratio vs. Risk Level')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Credit Score', y='Interest Rate', hue='Risk Level', data=df, palette=['#10b981', '#f59e0b', '#f43f5e'], alpha=0.7)
plt.title('Interest Rate vs. Credit Score (Colored by Risk)')
plt.show()
```

## 6. Correlation Heatmap (Numerical Features)

```python
plt.figure(figsize=(12, 10))
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5, cbar_kws={"shrink": .8})
plt.title('Correlation Matrix of Numerical Features', fontsize=14)
plt.show()
```
